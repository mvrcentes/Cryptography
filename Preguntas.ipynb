* Preguntas

* ¿Qué tamaño de clave se está usando para DES, 3DES y AES?
    * *DES*: Usa una clave de 56 bits (pero la implementación requiere 64 bits, donde 8 bits son de paridad y no aportan seguridad).
	* *3DES*: Usa claves de 112 bits (16 bytes) o 168 bits (24 bytes), dependiendo si se usa 2 o 3 claves DES.
	* *AES*: Puede usar claves de 128 bits (16 bytes), 192 bits (24 bytes) o 256 bits (32 bytes).
* ¿Qué modo de operación está implementado?
	* *DES*: ECB (Electronic Codebook)
	* *3DES*: CBC (Cipher Block Chaining)
	* *AES*: ECB y CBC
* ¿Por qué no debemos usar ECB en datos sensibles?
	1. ECB cifra cada bloque de datos de manera independiente. Esto significa que si hay partes iguales en el mensaje original, seguirán siendo iguales en el mensaje cifrado. En imágenes, por ejemplo, los patrones pueden seguir siendo visibles, lo que permite a alguien reconocer parte de la información.

* ¿Cual es la diferencia entre ECB vs CBC, se puede notar directamente en una imagen?
	* La diferencia principal entre ECB y CBC es la manera en que manejan los bloques de datos. ECB cifra cada bloque de forma independiente, sin mezclarlo con los anteriores, CBC usa un Vector de Inicialización (IV) y mezcla cada bloque con el anterior antes de cifrarlo. Esto hace que los datos cifrados sean más aleatorios y evita que se conserven patrones visibles.
* ¿Que es el IV?
	* El IV (Vector de Inicialización) es un valor aleatorio que se usa al comienzo del cifrado en ciertos modos como CBC. Su propósito es asegurarse de que, aunque dos mensajes sean iguales, el resultado cifrado sea diferente cada vez.
* ¿Que es el PADDING?
	* El Padding (relleno) se usa en cifrado cuando los datos no tienen el tamaño exacto requerido por el algoritmo. La mayoría de los cifrados requieren que los datos sean múltiplos de un cierto número de bytes, como 8 o 16. Si el mensaje no es del tamaño adecuado, se agregan bytes extra al final para completar el bloque.
* ¿En qué situaciones se recomienda cada modo de operación?
	* ECB no se recomienda para datos sensibles porque no oculta patrones en la información cifrada. Sin embargo, en casos donde la seguridad no es un problema, como bases de datos de valores fijos o pruebas, podría usarse.

	* CBC es una mejor opción cuando se necesita seguridad, como en el cifrado de archivos, bases de datos, imágenes y contraseñas. Su uso del IV y la forma en que mezcla los bloques lo hace mucho más seguro que ECB.
