# DES

# Configuración de Directorios

```python
import os

# Definir rutas de directorios
BASE_DIR = "Texts"
ENCRYPTED_DIR = os.path.join(BASE_DIR, "encrypted")
DECRYPTED_DIR = os.path.join(BASE_DIR, "decrypted")

# Crear los directorios si no existen
os.makedirs(ENCRYPTED_DIR, exist_ok=True)
os.makedirs(DECRYPTED_DIR, exist_ok=True)

print("Directorios creados/verificados correctamente.")
```

# Importar Librerías y Configurar DES

```python
from Crypto.Cipher import DES
import secrets
```

# Funciones de Relleno

```python
def pad_manual(text, block_size=8):
    """Agrega relleno manual para que el texto sea múltiplo de 8 bytes."""
    pad_len = block_size - (len(text) % block_size)
    padding = chr(pad_len) * pad_len  # Relleno con el número de bytes necesarios
    return text + padding

def unpad_manual(text):
    """Elimina el relleno manual después del descifrado."""
    pad_len = ord(text[-1])  # Último byte indica la cantidad de relleno
    return text[:-pad_len]  # Elimina los bytes de relleno
```

# Función para Generar Clave DES

```python
def generate_des_key():
    """Genera una clave DES aleatoria de 64 bits (8 bytes)."""
    return secrets.token_bytes(8)

# Generar clave
key = generate_des_key()
print(f"Clave DES generada (hex): {key.hex()}")
```

# Funciones de Cifrado y Descifrado

```python
def des_encrypt_ecb(plaintext, key):
    """Cifra un mensaje con DES en modo ECB."""
    cipher = DES.new(key, DES.MODE_ECB)
    padded_text = pad_manual(plaintext)  # Aplicar padding manual
    ciphertext = cipher.encrypt(padded_text.encode())  # Convertir a bytes antes de cifrar
    return ciphertext.hex()  # Convertimos a hexadecimal para facilidad de almacenamiento

def des_encrypt_ecb(plaintext_bytes, key):
    """Cifra un mensaje con DES en modo ECB."""
    cipher = DES.new(key, DES.MODE_ECB)
    ciphertext = cipher.encrypt(plaintext_bytes)  # Encrypt directly without extra padding
    return ciphertext  # Return raw encrypted bytes
```

# Función para Guardar y Leer Archivos

```python
def save_text(data, filename, folder):
    """Guarda un archivo en la carpeta especificada."""
    path = os.path.join(folder, filename)
    with open(path, "w") as file:
        file.write(data)
    print(f"Archivo guardado en: {path}")

def read_text(filename, folder):
    """Lee un archivo desde la carpeta especificada."""
    path = os.path.join(folder, filename)
    with open(path, "r") as file:
        return file.read()
```

#  Cifrar un Archivo con DES

```python
from Crypto.Util.Padding import pad
from Crypto.Cipher import DES

# Leer el texto desde el archivo des.txt
plaintext = read_text("des.txt", BASE_DIR).strip()  # Remover espacios en blanco extra

# Convertir texto a bytes y aplicar padding
plaintext_bytes = pad(plaintext.encode(), DES.block_size)  # Aplicar padding automático

# Cifrar el mensaje
encrypted_bytes = des_encrypt_ecb(plaintext_bytes, key)

# Convertir a hexadecimal antes de guardar
encrypted_text = encrypted_bytes.hex()

# Guardar en archivo
save_text(encrypted_text, "des_encrypted.txt", ENCRYPTED_DIR)

print(f"Texto cifrado (hex): {encrypted_text}")
```

# Descifrar un Archivo con DES

```python
# Leer el archivo cifrado
encrypted_text = read_text("des_encrypted.txt", ENCRYPTED_DIR)

# Descifrar el mensaje
decrypted_text = des_decrypt_ecb(encrypted_text, key)

# Guardar en archivo
save_text(decrypted_text, "des_decrypted.txt", DECRYPTED_DIR)

print(f"Texto descifrado: {decrypted_text}")
```
