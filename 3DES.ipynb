# 3DES

# Configuración de Directorios

```python
import os

# Definir rutas de directorios
BASE_DIR = "Texts"
ENCRYPTED_DIR = os.path.join(BASE_DIR, "encrypted")
DECRYPTED_DIR = os.path.join(BASE_DIR, "decrypted")

# Crear los directorios si no existen
os.makedirs(ENCRYPTED_DIR, exist_ok=True)
os.makedirs(DECRYPTED_DIR, exist_ok=True)

print("Directorios creados/verificados correctamente.")
```

# Importar Librerías y Configurar 3DES

```python
from Crypto.Cipher import DES3
import secrets
```

# Funciones de Relleno

```python
def pad_manual(text, block_size=8):
    """Agrega relleno manualmente para que el texto sea múltiplo de block_size (PKCS5 Padding)."""
    pad_len = block_size - (len(text) % block_size)
    padding = chr(pad_len) * pad_len
    return text + padding

def unpad_manual(text):
    """Elimina el relleno manual después del descifrado."""
    pad_len = ord(text[-1])
    return text[:-pad_len]
```

# Generar Clave y IV para 3DES

```python
def generate_3des_key():
    """Genera una clave aleatoria de 16 o 24 bytes para 3DES."""
    key_length = 24
    return secrets.token_bytes(key_length)

def validate_3des_key(key):
    """Verifica que la clave tenga exactamente 16 o 24 bytes."""
    if len(key) not in [16, 24]:
        raise ValueError("La clave 3DES debe tener exactamente 16 o 24 bytes.")

def generate_iv():
    """Genera un IV aleatorio de 8 bytes para 3DES en modo CBC."""
    return secrets.token_bytes(8)

# Generamos clave y IV
key = generate_3des_key()
iv = generate_iv()

print(f"Clave 3DES generada (hex): {key.hex()}")
print(f"IV 3DES generado (hex): {iv.hex()}")
```

# Funciones de Cifrado y Descifrado con 3DES

```python
def triple_des_encrypt_cbc(plaintext, key, iv):
    """Cifra un mensaje con 3DES en modo CBC."""
    validate_3des_key(key)
    cipher = DES3.new(key, DES3.MODE_CBC, iv)
    padded_text = pad_manual(plaintext)
    ciphertext = cipher.encrypt(padded_text.encode())
    return ciphertext.hex()

def triple_des_decrypt_cbc(ciphertext, key, iv):
    """Descifra un mensaje cifrado con 3DES en modo CBC."""
    validate_3des_key(key)
    cipher = DES3.new(key, DES3.MODE_CBC, iv)
    decrypted_text = cipher.decrypt(bytes.fromhex(ciphertext))
    return unpad_manual(decrypted_text.decode())
```

# Funciones para Guardar y Leer Archivos

```python
def save_text(data, filename, folder):
    """Guarda un archivo en la carpeta especificada."""
    path = os.path.join(folder, filename)
    with open(path, "w") as file:
        file.write(data)
    print(f"Archivo guardado en: {path}")

def read_text(filename, folder):
    """Lee un archivo desde la carpeta especificada."""
    path = os.path.join(folder, filename)
    with open(path, "r") as file:
        return file.read()
```

# Cifrar un Archivo con 3DES

```python
# Leer el texto desde el archivo 3des.txt
plaintext = read_text("3des.txt", BASE_DIR).strip()

# Cifrar el mensaje
encrypted_text = triple_des_encrypt_cbc(plaintext, key, iv)

# Guardar en archivo
save_text(encrypted_text, "3des_encrypted.txt", ENCRYPTED_DIR)

print(f"Texto cifrado (hex): {encrypted_text}")
```

# Descifrar un Archivo con 3DES

```python
# Leer el archivo cifrado
encrypted_text = read_text("3des_encrypted.txt", ENCRYPTED_DIR)

# Descifrar el mensaje
decrypted_text = triple_des_decrypt_cbc(encrypted_text, key, iv)

# Guardar en archivo
save_text(decrypted_text, "3des_decrypted.txt", DECRYPTED_DIR)

print(f"Texto descifrado: {decrypted_text}")
```
